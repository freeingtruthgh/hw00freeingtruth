```python
from freeingtruth.differential.discrete import diff
import matplotlib.pyplot as plt

t = [0, 0.1, 0.3, 0.4, 0.55, 0.67, 0.71]
x = [23.1, 22.5, 23.5, 21.88, 22.5, 23.5, 24.88]
v = diff(t,x)

print(v)

# Create plot with circle markers

# plot signal vs time
plt.plot(t,x,marker='o',linewidth=2)
# create title, and x, y labels
plt.title("Signal vs Time")
plt.xlabel("Time [s]")
plt.ylabel("Signal")
plt.savefig("Signal_vs_Time.pdf")    # save PDF of plot
plt.show()

# plot derivative vs time
plt.plot(t[1:],v,marker='o',linewidth=2)
# create title, and x, y labels
plt.title("Derivative vs Time")
plt.xlabel("Time [s]")
plt.ylabel("Derivative")
plt.savefig("Signal_vs_Time.pdf")    # save PDF of plot
plt.show()
```
